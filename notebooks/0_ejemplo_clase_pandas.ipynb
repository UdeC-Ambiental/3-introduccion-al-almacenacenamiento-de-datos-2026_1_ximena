# 🧩 Ejemplo de clase – Introducción a Pandas

Aprenderás los conceptos básicos de `pandas`, la creación de DataFrames y su exploración inicial.

```python
import pandas as pd

# Crea un DataFrame de ejemploo
datos = pd.DataFrame({
    'sitio': ['Río A', 'Río B', 'Río C'],
    'pH': [7.1, 8.3, 6.2]
})

datos.head()
```

# Reflexión
Este código crea una tabla (DataFrame) con datos de pH en diferentes ríos usando pandas, lo que permite organizar y visualizar la información fácilmente.

```python
datos.describe()
```

# Explicación corta
datos.describe() genera un resumen estadístico de las variables numéricas del DataFrame, facilitando el análisis rápido de los datos.
