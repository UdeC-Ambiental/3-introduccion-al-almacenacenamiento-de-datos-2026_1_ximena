**Objetivo:** Aplicar funciones estadísticas en Pandas para analizar datos, calcular promedios y crear nuevas columnas usando apply() y groupby().

```python
import pandas as pd
datos = pd.read_csv('../data/4_resultados_agua.csv')
```

### 🔹 Ejemplo de función personalizada
Crea una función que clasifique el pH:

```python
def clasificar_ph(valor):
    if valor < 6.5:
        return 'Ácido'
    elif valor <= 8.5:
        return 'Neutro'
    else:
        return 'Básico'

datos['categoria_ph'] = datos['pH'].apply(clasificar_ph)
datos.head()
```

```python
print(datos[['pH', 'categoria_ph']].head())


```

```python
datos.head()
```

```python
promedio_ph = datos['pH'].mean()
promedio_od = datos['OD_promedio'].mean()
promedio_temp = datos['Temp'].mean()

print("Promedio pH:", promedio_ph)
print("Promedio OD:", promedio_od)
print("Promedio Temperatura:", promedio_temp)

```

```python
promedios_categoria = datos.groupby('categoria_ph').mean(numeric_only=True)

for categoria, valores in promedios_categoria.iterrows():
    print(f"\nCategoría: {categoria}")
    print(valores)
```

📄 Análisis:

Los resultados muestran que los datos fueron agrupados según la categoría de pH (ácido, neutro y básico). Se observa que cada grupo presenta diferencias en los promedios de pH, oxígeno disuelto (OD) y temperatura.

En general, las aguas neutras suelen presentar valores más equilibrados, mientras que las ácidas o básicas pueden mostrar variaciones en las demás variables. Esto permite identificar posibles condiciones de calidad del agua y su comportamiento en cada categoría.
